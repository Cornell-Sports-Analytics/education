## What are NumPy and pandas?

NumPy and pandas are common Python libraries for data science. A library is a collection of prewritten code that helps us complete common tasks without starting from scratch.

| Library | Purpose | Example |
|---|---|---|
| **NumPy** | Performing calculations with numbers | Find the average or highest point total |
| **pandas** | Working with tables of data | View players, select columns, and filter games |

Both libraries store data in unique but similar ways. In Numpy, data is stored in a arrays which are just lists of numbers. In pandas, data is stored in DataFrames which are tables with rows and columns.

## 6. Calculating statistics (player avgs.)

Before calculating scoring statistics, we will keep rows where the player recorded more than zero minutes. This removes players who did not enter the game. When working with datasets, it is important to remove any null data or rows that do not fit the question you are trying to answer.

```python
played = df[df["numMinutes"] > 0]
```

The dataset has one row for each player-game. To find each player's season averages, we use `.groupby()` to place all rows belonging to the same player into one group. We then select the `points` and `assists` columns and use `.mean()` to calculate each player's averages.

```python
player_averages = played.groupby(["firstName", "lastName"])[["points", "assists"]].mean()
player_averages.head()
```

Next, we convert the `points` column into a NumPy array. Then, we use NumPy to find the average and the highest point total across the league.

```python
points = played["points"].to_numpy()

average_points = np.mean(points)
highest_points = np.max(points)

print("Average points:", round(average_points, 2))
print("Highest point total:", highest_points)
```

## Preview for NME #2: Visualizing Data with Matplotlib

Next week, we will learn how to turn data into visualizations and graphs. We will leverage Matplotlib, another Python library. For a quick preview, I've created a simple scatterplot below that uses `player_averages` to compare each player's average assists and average points.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.scatter(
    player_averages["assists"],
    player_averages["points"],
)

plt.title("Average Assists vs. Average Points")
plt.xlabel("Average Assists per Game")
plt.ylabel("Average Points per Game")

plt.show()
```
